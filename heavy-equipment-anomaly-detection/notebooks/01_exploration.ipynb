###Data Exploration

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

DB_PATH = "/content/machine_telemetry.db"

with sqlite3.connect(DB_PATH) as conn:
    dataframe_machine = pd.read_sql_query(
        "SELECT * FROM machine_telemetry",
        conn
    )

dataframe_machine.head()
dataframe_machine.info()
dataframe_machine.describe()

dataframe_machine["operating_state"].value_counts()
dataframe_machine["operating_state"].value_counts(normalize=True)

dataframe_machine.plot(
    x="timestamp",
    y="engine_load",
    figsize=(16, 5),
    title="Engine Load"
)

plt.ylabel("Load")
plt.show()

dataframe_machine.plot(
    x="timestamp",
    y="engine_temperature",
    figsize=(16, 5),
    title="Engine Temperature"
)

plt.ylabel("Temperature [°C]")
plt.show()

dataframe_machine.plot(
    x="timestamp",
    y="vibration",
    figsize=(16, 5),
    title="Machine Vibration"
)

plt.ylabel("Vibration")
plt.show()

numeric_columns = [
    "engine_load",
    "engine_rpm",
    "engine_temperature",
    "hydraulic_pressure",
    "hydraulic_temperature",
    "vibration",
    "fuel_rate",
]

dataframe_machine[numeric_columns].corr()

plt.figure(figsize=(10, 8))

plt.imshow(
    dataframe_machine[numeric_columns].corr(),
    aspect="auto"
)

plt.xticks(
    range(len(numeric_columns)),
    numeric_columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(numeric_columns)),
    numeric_columns
)

plt.colorbar(label="Correlation")

plt.title("Sensor Correlation Matrix")
plt.tight_layout()
plt.show()

dataframe_machine.groupby("operating_state")[numeric_columns].mean().round(2)
```
